```python
import chromadb

client = chromadb.PersistentClient(path="data/chroma_test")

collection = client.create_collection(name="chunks", embedding_function=None)
```

```python
collection.add(
    ids=["id1", "id2", "id3"],
    embeddings=[[1.1, 2.3, 3.2], [4.5, 6.9, 4.4], [1.1, 2.3, 3.2]],
    documents=["doc1", "doc2", "doc3"],
    metadatas=[{"chapter": 3, "verse": 16}, {"chapter": 3, "verse": 5}, {"chapter": 29, "verse": 11}],
)
```

```python
collection.update(ids=["id1"], metadatas=[])
```

```python
import pandas as pd

df = pd.DataFrame(collection.get(ids=["id1", "id2"]))
```
