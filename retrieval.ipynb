```python
from ddd_policy_tracer.enrichment.embedding import EmbeddingService
from ddd_policy_tracer.retrieval.sqlite_retrieval import SQLiteHybridRetriever
from pathlib import Path

embedding = EmbeddingService()
db_path = Path("data/new3.db")
retriever = SQLiteHybridRetriever(db_path)
```

```python
import typing

query = "what have people said about immigration policies?"
query_embedding = typing.cast(list[float], embedding.embed_one(query, prompt_name="query"))

results = retriever.search_hybrid(
    query=query,
    query_embedding=query_embedding,
)
```
